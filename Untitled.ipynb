```python
# Read the PDF


from pypdf import PdfReader

pdf_path = "/Users/vishnu/Downloads/IBMDATASTAGE_BOOK.pdf"
reader = PdfReader(pdf_path)

full_text = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        full_text += text + "\n"
```

```python
# Chuking the text

def chunk_text(text, chunk_size=800, overlap=100):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap
    return chunks

chunks = chunk_text(full_text)
print(len(chunks))
```

```python
from transformers import AutoTokenizer, AutoModel
import torch

EMB_MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

tokenizer = AutoTokenizer.from_pretrained(EMB_MODEL_NAME)
model = AutoModel.from_pretrained(EMB_MODEL_NAME)

def embed_texts(texts):
    inputs = tokenizer(
        texts,
        padding=True,
        truncation=True,
        max_length=256,
        return_tensors="pt",
    )
    with torch.no_grad():
        outputs = model(**inputs)
        last_hidden = outputs.last_hidden_state          # [batch, seq, dim]
        mask = inputs["attention_mask"].unsqueeze(-1).float()
        pooled = (last_hidden * mask).sum(dim=1) / mask.sum(dim=1)  # mean pooling
    return pooled.numpy()
```

```python
chunk_embeddings = embed_texts(chunks)
print(chunk_embeddings.shape)
```

```python
import faiss
import numpy as np

embeddings = np.array(chunk_embeddings).astype("float32")
index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)
print(index.ntotal)
```

```python
def retrieve_context(query, top_k=3):
    # use the new embedding function, not embed_model
    query_vec = embed_texts([query]).astype("float32")
    distances, indices = index.search(query_vec, top_k)
    return [chunks[i] for i in indices[0]]
```

```python
import requests

MODEL_NAME = "gemma3:1b"

def ask_ollama(prompt: str, max_tokens: int = 200) -> str:
    url = "http://localhost:11434/api/generate"
    payload = {
        "model": MODEL_NAME,
        "prompt": prompt,
        "stream": False,
        "options": {
            "num_predict": max_tokens,
            "temperature": 0.2,
            "top_p": 0.9
        }
    }
    resp = requests.post(url, json=payload)
    resp.raise_for_status()
    return resp.json()["response"]

def ask_book(question):
    context_chunks = retrieve_context(question, top_k=3)
    context = "\n\n".join(context_chunks)
    prompt = f"""
Answer only from the provided context.
If the answer is not in the context, say "I could not find it in the DataStage book."

Context:
{context}

Question:
{question}
"""
    return ask_ollama(prompt, max_tokens=180)
```

```python
while True:
    q = input("Ask about DataStage (or 'exit'): ")
    if q.strip().lower() in {"exit", "quit"}:
        break
    answer = ask_book(q)
    print("\nAnswer:\n")
    print(answer)
    print("\n" + "-" * 50 + "\n")
```
